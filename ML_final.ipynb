# 5.Dataset description and visualisation

| Property | BNCI 2014-008, "P300 speller with ALS patients" |
|---|---|
| Source | Riccio et al. (2013), *Frontiers in Human Neuroscience* 7:732; BNCI Horizon 2020 database; loaded with MOABB |
| Participants | 8 people with amyotrophic lateral sclerosis (ALS) |
| Task | copy-spelling of 7 predefined five-letter words = 35 characters per participant |
| Interface | 6 × 6 row/column P300 speller (A–Z, 1–9, `_`) |
| Stimulation | rows and columns flash in random order, 125 ms flash + 125 ms pause (4 Hz); 12 flashes (6 rows + 6 columns) = one *sequence*; 10 sequences per character |
| EEG | 8 electrodes Fz, Cz, Pz, Oz, P3, P4, PO7, PO8 (10-10 system), reference right earlobe, ground left mastoid, g.MOBIlab amplifier |
| Sampling | 256 Hz, amplifier band-pass 0.1–30 Hz |
| Type of data | continuous multichannel EEG time series (voltage) with a marker for every flash; after preprocessing, one epoch (8 channels × 128 samples) per flash |
| Samples | 35 × 10 × 12 = 4,200 flashes per participant, 33,600 in total |
| Classes | per flash: **target** (the flashed row/column contains the attended character) vs **non-target**, 700 vs 3,500 (1 : 5). Per half-sequence (6 flashes): which row or column is the target, 6 classes. Per character: 1 of 36 |
| Train / test | first 7 characters (840 flashes) train, remaining 28 characters (3,360 flashes) test |

**Preprocessing done by the cells above** (the authors' `get_processed_data`): 0.1–24 Hz Butterworth band-pass (zero phase); epochs from −0.3 to +1.0 s around each flash with baseline correction; resampling to 128 Hz (167 samples); dropping the first 39 samples (39/128 Hz = 0.305 s), so the 128 kept samples cover about 0–1 s after the flash; conversion to µV. `X` is ordered as [character, sequence, half (0 = columns, 1 = rows), row/column, channel, time].

The cells below describe participant 1 (the `X`, `y` loaded above) and then all 8 participants. They only read the data; nothing used by the model is changed.

# 14.Training and saving all 8 patient:

```python
for subject_id in range(1, 9):

    print(f"\n{'='*50}")
    print(f"Training GLASS for subject {subject_id}")
    print(f"{'='*50}")

    # -------------------------
    # 1. Load and preprocess data
    # -------------------------
    X, y = data.get_processed_data(
        subject_id=subject_id,
        sfreq=128,
        fmin=0.1,
        fmax=24
    )

    X = X[:, :, :, :, :, 39:]
    X *= 1e6

    n_channel, n_time = X.shape[-2:]

    # -------------------------
    # 2. Train/test split
    # -------------------------
    X_train, y_train = X[:7], y[:7]
    X_test, y_test = X[7:], y[7:]

    # -------------------------
    # 3. Reshape for GLASS
    # -------------------------
    X_train_glass = X_train.reshape(
        [-1, 6, n_channel, n_time]
    )

    y_train_glass = y_train.reshape(
        [-1, 6]
    )

    # -------------------------
    # 4. First model: no shrinkage
    # -------------------------
    model = glass.Glass()

    model.process_data(
        X_train_glass,
        y_train_glass
    )

    model.mfvb(
        2000,
        learning_rate=0.05,
        seed=1
    )

    # -------------------------
    # 5. Calculate cutoff
    # -------------------------
    cutoff = np.median(
        np.abs(model.betaMat)
    )

    print(f"Cutoff = {cutoff}")

    # -------------------------
    # 6. Final model: shrinkage
    # -------------------------
    model = glass.Glass(
        shrinkage_factor=0.5 * cutoff
    )

    model.process_data(
        X_train_glass,
        y_train_glass
    )

    model.mfvb(
        2000,
        learning_rate=0.05,
        seed=1
    )

    # -------------------------
    # 7. Predict on test data
    # -------------------------
    yhat_test = model.predict_prob(
        X_test.reshape(
            [-1, 6, n_channel, n_time]
        )
    )

    # -------------------------
    # 8. Restore test shape
    # -------------------------
    yhat_test = yhat_test.reshape(
        y_test.shape
    )

    # -------------------------
    # 9. Evaluate
    # -------------------------
    result = helper_funcs.evaluate(
        yhat_test,
        y_test
    )

    # -------------------------
    # 10. Save to Google Drive
    # -------------------------
    output_path = (
        f'/content/drive/MyDrive/'
        f'GLASS_results/subject_{subject_id}.json'
    )

    with open(output_path, 'w') as f:
        json.dump(result, f)

    print(f"Saved: {output_path}")
    print(f"Final accuracy: {result['accuracy'][-1]:.4f}")

    # ---- ADDED for the Results section (requirement 3): save predictions and model summaries ----
    yhat_vote = model.predict_prob(X_test.reshape([-1, 6, n_channel, n_time]), method='vote').reshape(y_test.shape)
    np.savez(
        f'/content/drive/MyDrive/GLASS_results/subject_{subject_id}_details.npz',
        yhat=yhat_test, yhat_vote=yhat_vote, y=y_test,
        scores=np.tensordot(X_test, model.betaMat, axes=[[4, 5], [0, 1]]),   # score of every test flash
        cutoff=cutoff, losses=np.array(model.losses),
        effective_weight=model.effective_weight,
        selection_prob=(model.probs.argmax(axis=2) != 0).mean(axis=0),
        beta_median=model.betagMat,
        beta_q025=np.percentile(model.betagMats, 2.5, axis=0),
        beta_q975=np.percentile(model.betagMats, 97.5, axis=0),
        beta_p_nonzero=(model.betagMats != 0).mean(axis=0))
```

# 15. Results and Performance Metrics

This section uses the files saved by the 8-participant loop above (`subject_k.json` and `subject_k_details.npz`). Section 3 retrains GLASS for the ablation study and Section 4 for the hyperparameter tuning; every finished participant is saved, so re-running a cell skips work already done.



| Metric | Definition |
|---|---|
| Character accuracy after *k* sequences | share of the 28 test characters whose row **and** column are both correct after *k* sequences (the paper's metric; chance = 1/36) |
| Single-sequence accuracy | share of half-sequences whose target row/column is found from **one** sequence (chance = 1/6) |
| Flash ROC AUC | how well the score η separates target from non-target flashes, without a threshold (0.5 = chance, 1 = perfect) |
| BCI utility (bits/min) | $(2p-1)\log_2 35 / c$, with $c$ = minutes per character ($k$ × 12 × 0.25 s), 0 when $p < 0.5$ (the paper's metric) |
| Wolpaw ITR (bits/min) | $[\log_2 N + p\log_2 p + (1-p)\log_2\frac{1-p}{N-1}]/c$, N = 36 (the most common BCI throughput metric) |

SDs use the sample standard deviation (`ddof=1`), as in the paper.

```python
import os, json, time, warnings, functools
from IPython.display import display, Markdown
from sklearn.metrics import roc_auc_score, roc_curve

RES_DIR = '/content/drive/MyDrive/GLASS_results'        # same folder as the training loop
AUTHORS_DIR = '/content/code/GLASS-main/output'          # authors' saved results (from code.zip)
SUBJECTS = ALL_SUBJECTS
NUM_STEPS = 2000
PAPER_S1 = {4: ((63.8, 20.8), (8.47, 8.97)), 7: ((77.2, 17.7), (7.98, 5.19)), 10: ((85.3, 13.0), (7.24, 2.67))}

def bci_utility(p, k):
    return np.clip((2 * np.asarray(p) - 1) * np.log2(35) / (k * 12 * 0.25 / 60), 0, None)

def wolpaw_itr(p, k, N=36):
    p = np.clip(np.asarray(p, float), 1e-12, 1)
    bits = np.log2(N) + p * np.log2(p) + (1 - p) * np.log2(np.clip((1 - p) / (N - 1), 1e-300, None))
    bits = np.where(p <= 1 / N, 0, np.where(p >= 1, np.log2(N), bits))
    return bits * 60 / (k * 12 * 0.25)

def ms(v, scale=1, nd=1):
    v = np.asarray(v, float) * scale
    return f"{v.mean():.{nd}f} ({v.std(ddof=1) if len(v) > 1 else float('nan'):.{nd}f})"

acc = {s: np.array(json.load(open(f"{RES_DIR}/subject_{s}.json"))["accuracy"]) for s in SUBJECTS}
det = {s: dict(np.load(f"{RES_DIR}/subject_{s}_details.npz")) for s in SUBJECTS}
auth = {s: np.array(json.load(open(f"{AUTHORS_DIR}/subject_{s}.json"))["accuracy"]) for s in SUBJECTS}
A_ = np.array([acc[s] for s in SUBJECTS]); K = np.arange(1, 11)

display(Markdown("**Supplementary Table S1 (GLASS row): mean (SD) over participants**"))
display(pd.DataFrame([{"sequences": k, "accuracy % (this notebook)": ms(A_[:, k - 1], 100),
                       "accuracy % (paper)": "%.1f (%.1f)" % PAPER_S1[k][0],
                       "BCI utility bits/min (this notebook)": ms(bci_utility(A_[:, k - 1], k), nd=2),
                       "BCI utility bits/min (paper)": "%.2f (%.2f)" % PAPER_S1[k][1]} for k in (4, 7, 10)]).set_index("sequences"))

display(Markdown("**Difference from the authors' saved results** (correct test characters, yours − authors', after 1..10 sequences)"))
display(pd.DataFrame({f"{k} seq": [int(round(28 * (acc[s][k - 1] - auth[s][k - 1]))) for s in SUBJECTS] for k in K},
                     index=[f"participant {s}" for s in SUBJECTS]))

def single_seq_acc(yh, yy): return float(np.mean(yh.argmax(3) == yy.argmax(3)))
def product_rule_acc(yh, yy):
    c = np.log(np.clip(yh, 1e-300, None)).cumsum(1).argmax(3) == yy.argmax(3)
    return (c.sum(2) == 2).mean(0)

rows = []
for s in SUBJECTS:
    d = det[s]; r = {"participant": s, "single-sequence acc %": 100 * single_seq_acc(d["yhat"], d["y"]),
                     "flash AUC": roc_auc_score(d["y"].ravel(), d["scores"].ravel())}
    for k in (4, 7, 10):
        r[f"acc % @{k}"] = 100 * acc[s][k - 1]; r[f"ITR @{k}"] = float(wolpaw_itr(acc[s][k - 1], k))
        r[f"utility @{k}"] = float(bci_utility(acc[s][k - 1], k))
    r["τ = 0.5 × cutoff"] = 0.5 * float(d["cutoff"]); rows.append(r)
metrics = pd.DataFrame(rows).set_index("participant")
metrics.loc["mean"] = metrics.mean(); metrics.loc["SD"] = metrics.iloc[:-1].std(ddof=1)
display(Markdown("**All metrics per participant**")); display(metrics.round(3))
```

## 17. Ablation study

Each variant removes or replaces **one** part of GLASS and keeps everything else (data, split, seed, two-fit τ rule) the same as in the loop above.

| Variant | What changes | Question it answers |
|---|---|---|
| A1 | no soft-thresholding (τ = 0, i.e. only fit 1) | does sparsity in time help? |
| A2 | prediction averaged over the 5000 posterior samples instead of the posterior median | does using the whole posterior help? |
| A3 | product of probabilities over sequences (the paper's rule) instead of the sum (the code) | does the combination rule matter? |
| A4 | 1 latent channel (the paper's description) instead of 2 | is a second spatial pattern useful? |
| A5 | only PO7, PO8, Oz, Pz (the paper's GLASSel) | can 4 electrodes replace 8? |
| A6 | independent binary (Bernoulli) likelihood with an intercept instead of the softmax over 6 | does modelling "one target among 6" help (the paper's central claim)? |
| A7, A8 | response window 0–550 ms / 0–350 ms instead of 0–1000 ms | where in time is the information? |

A2 and A3 reuse the saved predictions. The others retrain: A1 needs 1 fit per participant, A4–A8 need 2 each, so 11 fits per participant in total. That is roughly 5–6 times as long as the 8-participant loop above. To save time, shorten `ABLATIONS` or `SUBJECTS` in the next cell.

```python
import tensorflow as tf
import tensorflow_probability as tfp
tfd, tfb = tfp.distributions, tfp.bijectors

ABLATIONS = ["no_threshold", "rank1", "sel4", "binary", "win550", "win350"]
SEL4 = ["PO7", "PO8", "Oz", "Pz"]


class GlassVariant(glass.Glass):
    '''The authors' Glass with a configurable number of latent channels (n_latent) and
    likelihood ("multinomial" = authors' model, "bernoulli" = independent binary classifier).
    With n_latent=2 and "multinomial" it is identical to glass.Glass.'''
    def __init__(self, shrinkage_factor=0, n_latent=2, likelihood="multinomial", dtype=tf.float32):
        super().__init__(shrinkage_factor, dtype)
        self.n_latent, self.likelihood = n_latent, likelihood

    def process_data(self, X_train, y_train):
        self.nchannel, self.nT = X_train.shape[2:]
        self.X_train = tf.constant(X_train, dtype=self.dtype)
        self.y_train = tf.constant(y_train, dtype=self.dtype)
        L = self.n_latent
        def jointmodel():
            sigma = yield tfd.Sample(tfd.HalfCauchy(0.0, tf.cast(1, self.dtype)), L)
            beta = yield tfb.Cumsum()(tfd.Sample(tfd.Normal(0.0, 1.0), (L, self.nT)))
            beta = tfp.math.soft_threshold(sigma[:, None] * beta, self.shrinkage_factor)
            probs = yield tfd.RelaxedOneHotCategorical(1, logits=tf.zeros([self.nchannel, L + 1]))
            weights = yield tfd.Sample(tfd.Normal(0.0, 1.0), [self.nchannel, L])
            weights = tf.linalg.normalize(weights, axis=0)[0]
            beta = tf.linalg.matmul(probs[:, 1:] * weights, beta)
            logits = tf.linalg.tensordot(self.X_train, beta, axes=[[2, 3], [0, 1]])
            if self.likelihood == "multinomial":
                y = yield tfd.Multinomial(logits=logits, total_count=1)
            else:
                b = yield tfd.Normal(tf.cast(0.0, self.dtype), tf.cast(10.0, self.dtype))
                y = yield tfd.Bernoulli(logits=logits + b, dtype=self.dtype)
        self.joint = tfd.JointDistributionCoroutineAutoBatched(jointmodel)

    def mfvb(self, num_steps=2000, sample_size=10, importance_sample_size=10, learning_rate=0.05, seed=1):
        L, dt = self.n_latent, self.dtype
        parts = [
            tfd.LogNormal(tf.Variable(tf.zeros(L, dtype=dt) - 3),
                          tfp.util.TransformedVariable(0.1 * tf.ones(L, dtype=dt), bijector=tfb.Softplus())),
            tfd.Normal(tf.Variable(tf.random.normal((L, self.nT), stddev=3 * self.shrinkage_factor, dtype=dt), dtype=dt),
                       tfp.util.TransformedVariable(0.01 * tf.ones((L, self.nT), dtype=dt), bijector=tfb.Softplus())),
            tfd.RelaxedOneHotCategorical(0.5, logits=tf.Variable(tf.zeros((self.nchannel, L + 1)))),
            tfd.Normal(tf.Variable(tf.zeros([self.nchannel, L], dtype=dt), dtype=dt),
                       tfp.util.TransformedVariable(tf.ones([self.nchannel, L], dtype=dt), dtype=dt, bijector=tfb.Softplus()))]
        if self.likelihood == "bernoulli":        # intercept, starting at log(1/5)
            parts.append(tfd.Normal(tf.Variable(tf.constant(-1.6, dtype=dt)),
                                    tfp.util.TransformedVariable(tf.constant(0.1, dtype=dt), bijector=tfb.Softplus())))
        self.posterior = tfd.JointDistributionSequentialAutoBatched(parts)
        optimizer = tf.optimizers.Adam(learning_rate=learning_rate)
        tf.random.set_seed(seed)
        self.losses = [float(x) for x in tfp.vi.fit_surrogate_posterior(
            self.loglik, self.posterior, optimizer=optimizer, num_steps=num_steps,
            sample_size=sample_size, importance_sample_size=importance_sample_size)]
        self.samples = self.posterior.sample(5000)
        self.process_samples()

    def process_samples(self):
        smp = [np.array(v) for v in self.samples]
        self.intercept = float(np.median(smp[4])) if self.likelihood == "bernoulli" else 0.0
        sigmas, self.betagMats, self.probs, self.weights = smp[:4]
        self.weights = self.weights / np.sqrt(np.sum(np.square(self.weights), 1))[:, None, :]
        self.effective_weights = self.probs[:, :, 1:] * self.weights
        self.weight, self.effective_weight = self.weights.mean(0), self.effective_weights.mean(0)
        self.betagMats = np.array(tfp.math.soft_threshold(sigmas[:, :, None] * self.betagMats, self.shrinkage_factor))
        self.betagMat = np.median(self.betagMats, axis=0)
        self.betaMats = np.matmul(self.effective_weights, self.betagMats)
        self.betaMat = np.median(self.betaMats, axis=0)

    def predict_prob(self, newX, method='median'):
        if self.likelihood == "multinomial":
            return super().predict_prob(newX, method)
        return 1 / (1 + np.exp(-(np.tensordot(newX, self.betaMat, axes=[[2, 3], [0, 1]]) + self.intercept)))


def fit_glass(Xtr, ytr, make=glass.Glass, ratio=0.5, num_steps=None, lr=0.05, seed=1, two_fits=True):
    '''The loop's procedure: fit 1 (tau = 0), cutoff = median|B|, fit 2 with tau = ratio x cutoff.'''
    num_steps = num_steps or NUM_STEPS
    nc, nt = Xtr.shape[-2:]; Xh, yh = Xtr.reshape([-1, 6, nc, nt]), ytr.reshape([-1, 6])
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        m = make(); m.process_data(Xh, yh); m.mfvb(num_steps, learning_rate=lr, seed=seed)
        cut = float(np.median(np.abs(m.betaMat)))
        if two_fits:
            m = make(shrinkage_factor=ratio * cut); m.process_data(Xh, yh); m.mfvb(num_steps, learning_rate=lr, seed=seed)
    return m, cut

def predict_and_score(m, Xte, yte):
    nc, nt = Xte.shape[-2:]
    yh = m.predict_prob(Xte.reshape([-1, 6, nc, nt])).reshape(yte.shape)
    sc = np.tensordot(Xte, m.betaMat, axes=[[4, 5], [0, 1]])
    return {"accuracy": helper_funcs.evaluate(yh, yte)["accuracy"], "auc": float(roc_auc_score(yte.ravel(), sc.ravel())),
            "single_seq_acc": single_seq_acc(yh, yte)}

ABL_SETUP = {   # name: (model, channels, window end in ms, two fits?)
    "no_threshold": (glass.Glass, None, None, False),
    "rank1": (functools.partial(GlassVariant, n_latent=1), None, None, True),
    "sel4": (glass.Glass, SEL4, None, True),
    "binary": (functools.partial(GlassVariant, likelihood="bernoulli"), None, None, True),
    "win550": (glass.Glass, None, 550, True),
    "win350": (glass.Glass, None, 350, True),
}
for name in ABLATIONS:
    make, chans, tmax, two = ABL_SETUP[name]
    for s in SUBJECTS:
        path = f"{RES_DIR}/ablation/{name}/subject_{s}.json"
        if os.path.exists(path):
            continue
        t0 = time.time(); Xs, ys = load_subject(s)
        Z = Xs
        if chans is not None:
            Z = Z[:, :, :, :, [CH.index(c) for c in chans], :]
        if tmax is not None:
            Z = Z[..., t_ms <= tmax]
        m, cut = fit_glass(Z[:7], ys[:7], make, two_fits=two)
        out = predict_and_score(m, Z[7:], ys[7:]); out.update({"cutoff": cut, "input_shape": list(Z[:7].shape)})
        os.makedirs(os.path.dirname(path), exist_ok=True); json.dump(out, open(path, "w"))
        print(f"{name:13s} participant {s}: accuracy at 10 seq = {100 * out['accuracy'][9]:.1f} %  ({time.time() - t0:.0f} s)")
print("ablation runs finished")
```



## 18. Hyperparameter tuning

**Shrinkage ratio** (τ = ratio × median|B|; the paper uses 0.5), tested for ratio ∈ {0, 0.25, 0.5, 1.0}:
- **Selection without using the test set:** inside the 7 training characters, fit on characters 1–5 and validate on characters 6–7. The criterion is the validation log-likelihood of the true targets. The best ratio is picked per participant and also globally (best mean).
- **Test-set sensitivity** of the same ratios (the paper's Table S2, settings 0, 2 and 3). It shows how much the choice matters, but is not used to choose. Ratio 0 is ablation A1 and 0.5 is the loop above.

**Optimiser settings** (Table S2, settings 4 and 5): 1000 steps with learning rate 0.1, and 4000 steps with learning rate 0.025, versus the default 2000 steps with 0.05.

Cost per participant: 4 short fits (validation), 2 fits (test ratios) and 4 fits (optimiser settings, one of them twice as long). That is roughly 5 times as long as the loop above; shorten `TUNING` or `SUBJECTS` to save time.

```python
# ============================================================
# HYPERPARAMETER TUNING
# ============================================================

TUNING = ["ratio", "optimizer", "mc_samples"]

# Use 3 subjects
SUBJECTS = [1, 2, 3]

# ------------------------------------------------------------
# Hyperparameter values
# ------------------------------------------------------------

# Shrinkage ratio
RATIOS = [0.0, 0.25, 0.5, 1.0]

# Optimizer / training schedules
SCHEDULES = {
    "S4: 1000 steps, lr 0.1": (1000, 0.1),
    "S5: 4000 steps, lr 0.025": (4000, 0.025)
}

# Monte Carlo sample size
MC_SAMPLES = [5, 10, 15]


# ============================================================
# Fixed-tau GLASS
# ============================================================

def fit_fixed_tau(Xtr, ytr, tau, num_steps=None):
    """
    Fit GLASS using a fixed shrinkage threshold tau.
    """
    nc, nt = Xtr.shape[-2:]

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        tf.random.set_seed(1)

        m = glass.Glass(
            shrinkage_factor=tau
        )

        m.process_data(
            Xtr.reshape([-1, 6, nc, nt]),
            ytr.reshape([-1, 6])
        )

        m.mfvb(
            num_steps=num_steps or NUM_STEPS,
            learning_rate=0.05,
            seed=1
        )

    return m


# ============================================================
# GLASS with variable Monte Carlo sample size
# ============================================================

def fit_glass_mc(
    Xtr,
    ytr,
    num_steps=None,
    lr=0.05,
    sample_size=10,
    seed=1
):
    """
    Fit GLASS with a specified Monte Carlo sample size.

    Everything else is kept fixed:
        - shrinkage factor = 0
        - learning rate = 0.05
        - importance sample size = 10
        - seed = 1
    """

    num_steps = num_steps or NUM_STEPS

    nc, nt = Xtr.shape[-2:]

    with warnings.catch_warnings():
        warnings.simplefilter("ignore")

        tf.random.set_seed(seed)

        m = glass.Glass()

        m.process_data(
            Xtr.reshape([-1, 6, nc, nt]),
            ytr.reshape([-1, 6])
        )

        m.mfvb(
            num_steps=num_steps,
            sample_size=sample_size,
            importance_sample_size=10,
            learning_rate=lr,
            seed=seed
        )

    return m


# ============================================================
# RUN TUNING
# ============================================================

for s in SUBJECTS:

    print(f"\n{'='*60}")
    print(f"Participant {s}")
    print(f"{'='*60}")

    # Load subject
    Xs, ys = load_subject(s)


    # ========================================================
    # 1. SHRINKAGE-RATIO TUNING
    # ========================================================

    path = f"{RES_DIR}/tuning/ratio/subject_{s}.json"

    if "ratio" in TUNING and not os.path.exists(path):

        t0 = time.time()

        # ----------------------------------------------------
        # Validation split
        # Trials 1-5 -> training
        # Trials 6-7 -> validation
        # ----------------------------------------------------

        Xa, ya = Xs[:5], ys[:5]
        Xv, yv = Xs[5:7], ys[5:7]

        # Fit tau = 0 model to obtain cutoff
        m0, cut_v = fit_glass(
            Xa,
            ya,
            two_fits=False
        )

        val = {}

        # Test each shrinkage ratio
        for r in RATIOS:

            if r == 0:

                m = m0

            else:

                m = fit_fixed_tau(
                    Xa,
                    ya,
                    r * cut_v
                )

            # Validation prediction
            nc, nt = Xv.shape[-2:]

            p = m.predict_prob(
                Xv.reshape([-1, 6, nc, nt])
            ).reshape(yv.shape)

            # Validation log-likelihood
            val_loglik = np.mean(
                np.log(
                    np.clip(
                        (p * yv).sum(-1),
                        1e-300,
                        None
                    )
                )
            )

            # Validation single-sequence accuracy
            val_acc = single_seq_acc(
                p,
                yv
            )

            val[str(r)] = {
                "val_loglik": float(val_loglik),
                "val_single_seq_acc": float(val_acc)
            }

        # Select best ratio based on validation log-likelihood
        best = max(
            RATIOS,
            key=lambda r: (
                val[str(r)]["val_loglik"],
                r == 0.5
            )
        )

        # ----------------------------------------------------
        # Test performance
        # Trials 1-7 -> training
        # Trials 8+  -> test
        # ----------------------------------------------------

        test = {
            "0.5": {
                "accuracy": acc[s].tolist()
            }
        }

        # Ratio = 0
        a1 = (
            f"{RES_DIR}/ablation/"
            f"no_threshold/subject_{s}.json"
        )

        if os.path.exists(a1):

            test["0.0"] = {
                "accuracy": json.load(
                    open(a1)
                )["accuracy"]
            }

        else:

            m = fit_glass(
                Xs[:7],
                ys[:7],
                two_fits=False
            )[0]

            test["0.0"] = {
                "accuracy": predict_and_score(
                    m,
                    Xs[7:],
                    ys[7:]
                )["accuracy"]
            }

        # Ratios = 0.25 and 1.0
        for r in (0.25, 1.0):

            m = fit_fixed_tau(
                Xs[:7],
                ys[:7],
                r * float(det[s]["cutoff"])
            )

            test[str(r)] = {
                "accuracy": predict_and_score(
                    m,
                    Xs[7:],
                    ys[7:]
                )["accuracy"]
            }

        # Save
        os.makedirs(
            os.path.dirname(path),
            exist_ok=True
        )

        json.dump(
            {
                "validation": val,
                "selected_ratio": best,
                "test": test
            },
            open(path, "w")
        )

        print(
            f"Ratio tuning participant {s} done "
            f"({time.time() - t0:.0f} s)"
        )


    # ========================================================
    # 2. OPTIMIZER / TRAINING-SCHEDULE TUNING
    # ========================================================

    path = (
        f"{RES_DIR}/tuning/"
        f"optimizer/subject_{s}.json"
    )

    if "optimizer" in TUNING and not os.path.exists(path):

        t0 = time.time()

        out = {}

        for name, (steps, lr) in SCHEDULES.items():

            # Keep the same scaling used in your original code
            m, _ = fit_glass(
                Xs[:7],
                ys[:7],
                num_steps=int(
                    steps * NUM_STEPS / 2000
                ),
                lr=lr
            )

            out[name] = predict_and_score(
                m,
                Xs[7:],
                ys[7:]
            )

        os.makedirs(
            os.path.dirname(path),
            exist_ok=True
        )

        json.dump(
            out,
            open(path, "w")
        )

        print(
            f"Optimizer tuning participant {s} done "
            f"({time.time() - t0:.0f} s)"
        )


    # ========================================================
    # 3. MONTE CARLO SAMPLE-SIZE TUNING
    # ========================================================

    path = (
        f"{RES_DIR}/tuning/"
        f"mc_samples/subject_{s}.json"
    )

    if "mc_samples" in TUNING and not os.path.exists(path):

        t0 = time.time()

        out = {}

        # ----------------------------------------------------
        # Keep everything else fixed:
        #
        # training trials  = first 7
        # test trials      = remaining trials
        # steps             = NUM_STEPS (2000)
        # learning rate     = 0.05
        # importance samples= 10
        # shrinkage         = 0
        # seed              = 1
        #
        # Only sample_size changes.
        # ----------------------------------------------------

        for mc in MC_SAMPLES:

            print(
                f"Participant {s}: "
                f"Monte Carlo samples = {mc}"
            )

            m = fit_glass_mc(
                Xs[:7],
                ys[:7],
                num_steps=NUM_STEPS,
                lr=0.05,
                sample_size=mc,
                seed=1
            )

            result = predict_and_score(
                m,
                Xs[7:],
                ys[7:]
            )

            out[str(mc)] = result

        # Save results
        os.makedirs(
            os.path.dirname(path),
            exist_ok=True
        )

        json.dump(
            out,
            open(path, "w")
        )

        print(
            f"Monte Carlo tuning participant {s} done "
            f"({time.time() - t0:.0f} s)"
        )


print("\n========================================")
print("ALL HYPERPARAMETER TUNING FINISHED")
print("========================================")
```
